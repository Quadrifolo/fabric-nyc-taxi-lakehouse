```python
# Vendor Performance

#Calculate the number of trips and average trip distance for each vendor.
```

```python
%%sql

SELECT
    VendorID,
    COUNT(*) AS NumberOfTrips,
    AVG(trip_distance) AS AverageTripDistance
FROM parquet.`Files/green_tripdata_2026-01.parquet`
GROUP BY VendorID
ORDER BY NumberOfTrips DESC
```
